```python
from deep_translator import GoogleTranslator

text = input("Enter text: ")

source = input("Source language code (en, hi, ml): ")

target = input("Target language code (en, hi, ml): ")

translated = GoogleTranslator(
    source=source,
    target=target
).translate(text)

print("\nTranslated Text:")
print(translated)
```
